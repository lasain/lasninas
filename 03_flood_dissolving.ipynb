```python
# Read the GeoPackage
import geopandas as gpd
import pathlib 
NOTEBOOK_PATH = pathlib.Path().resolve()
gdf = gpd.read_file(NOTEBOOK_PATH/"data" / "tulva_kohtalainen"/ "Vedensyvyys RCP4,5.shp")

# Check columns and CRS
print(gdf.columns)
print(gdf.crs)
print(len(gdf))
```

```python
# dissolve based on the depths
flood_dissolved = gdf.dissolve(
    by="DN",
    method="coverage",
    as_index=False
)
```

```python
# save to file
flood_dissolved.to_file(NOTEBOOK_PATH/
    "data"/"tulva_kohtalainen"/"flood_depths_moderate.gpkg",
    driver="GPKG")
```
